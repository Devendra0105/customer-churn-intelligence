*connect python script with database*

```python
import pandas as pd
import os
from sqlalchemy import URL, create_engine
from dotenv import load_dotenv
load_dotenv()
from pathlib import Path
```

```python
url=URL.create(
    drivername="postgresql+psycopg",
    username=os.getenv("DB_USER","postgres"),
    password=os.getenv("DB_PASSWORD"),
    host=os.getenv("DB_HOST","localhost"),
    port=int(os.getenv("DB_PORT",5432)),
    database=os.getenv("DB_NAME","churn_dataset")    
    
)
engine=create_engine(url)
print(engine)
```

```python
base=Path('Extract_from_DB.ipynb').resolve().parent.parent
try : 
    df=pd.read_sql_query("select * from public.telco_churn_raw",engine)
    print('Data Extracted Successfully')
finally:
    engine.dispose()
    print('Connection has successfully closed')
df.to_csv(base/'data'/'raw_data.csv',index=False)
```
